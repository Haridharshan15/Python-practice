```python
#palindrome number

def pal(n):
  temp=n
  c=0
  b=0
  while temp!=0:
    b=int(temp%10)
    temp=int(temp/10)
    #print(b," ",temp)
    #c=b+c
    c=(c*10)+b
    # print("in loop",c)
  #c=int(c/10)
  return "Pallindrome" if c==n else "Not Pallindrome"

n=1223221
#n=int(input("Enter a number:"))
print(pal(n))
```

```python
#amstrong number
def fib(n):
  temp=n
  c=0
  while temp!=0:
    b=int(temp%10)
    temp=int(temp/10)
    c=c+(b*b*b)
    b=0
  return "Amstrong" if c==n else "Not Amstrong"

n=153
print(fib(n))
```

```python
sentence = "The document you are colab not a static page, document that a colab that you write colab"
words = sentence.split()
word_count = len(words)
print("Total number of words:", word_count)

word_counts = {}
for word in words:
  word_counts[word] = word_counts.get(word, 0) + 1


print("Word counts:", word_counts)
```

```python
 def findClosest(x, y, z):
        if z>y or z>x:
            y=z-y
            x=z-x
        elif y>z or x>z:
            y=y-z
            x=x-z
        if x<y:
            return 1
        elif y<x:
            return 2
        else:
            return 0



print(findClosest(2,7,4))
```
